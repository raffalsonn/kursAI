Napisz program rozwiązający układ równań z 2 niewiadomymi metodą Cramera.
Wszystkie współczynniki powinny być podane przez użytkownika z klawiatury.
Wyświetlić wykres funkcji na odcinku x oraz przecięciu wykresu.

```python
import numpy as np
import matplotlib.pyplot as plt

x1, y1, score1 = map(float, input("podaj wspolczynniki x, y i wyraz wolny:").split())
x2, y2, score2 = map(float, input("podaj wspolczynniki x, y i wyraz wolny:").split())
A = np.array([[x1,y1],[x2,y2]],dtype=float)
B= np.array([score1,score2],dtype=float)
D = np.linalg.det(A)

if D!=0:
  A_x = A.copy()
  A_x[:,0] = B
  D_x = np.linalg.det(A_x)
  x = D_x/D
  x_vals = np.linspace(x-10,x+10,100)
  A_y = A.copy()
  A_y[:,1] = B
  D_y = np.linalg.det(A_y)
  y = D_y/D
  yvals1 = (score1 - x1*x_vals) / y1
  yvals2 = (score2 - x2*x_vals) / y2

  plt.plot(x_vals, yvals1,label="y1")
  plt.plot(x_vals, yvals2,label="y2")
  plt.scatter([x],[y], color='r')

plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Układ równań")
plt.show()
```
